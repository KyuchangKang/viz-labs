## Part A. 준비 — 다섯 명의 성적표

**V1**에서 쓰던 그 다섯 명입니다. V1 §5에서 **두 세트 막대**로 세워 본 중간·기말이
오늘은 **한 표의 두 열**로 들어 있습니다. 이 셀은 **그대로 실행만 하세요.**

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.DataFrame({
    "이름": ["강수빈", "노민준", "정하람", "오하린", "한도윤"],
    "학과": ["컴퓨터", "경영", "컴퓨터", "컴퓨터", "경영"],
    "중간": [88, 76, 82, 87, 83],
    "기말": [89, 80, 92, 91, 87],
})

print(df)
print("모양:", df.shape)
print("seaborn 판:", sns.__version__)
```

## Part B. seaborn 한 줄 — `data=`·`x=`·`y=`

값이 아니라 **열 이름을 글자로** 넘깁니다. `ax=`로 내가 만든 Axes에 얹으면
그다음은 전부 V1에서 배운 `ax.` 문법입니다. 빈칸 없음.

```python
fig1, ax1 = plt.subplots()
sns.barplot(data=df, x="이름", y="기말", ax=ax1)
ax1.set_title("Final scores")
ax1.set_xlabel("Student")            # 안 붙이면 판다스처럼 한글 열 이름이 자동으로 붙어 네모(□)가 된다
ax1.set_ylabel("Score")

print("막대 높이:", [round(float(p.get_height()), 1) for p in ax1.patches])
print("x 눈금:", [t.get_text() for t in ax1.get_xticklabels()])
plt.show()
```

x축 **눈금 글자**는 표의 값(이름)이라 한글 그대로이고 화면에는 네모로 보입니다 —
V4에서 못 박은 대로 **순서를 확인하기 전에는 눈금 글자를 갈아 끼우지 않습니다.**

## Part C. 세트피스 ⭐ — 무리를 가르는 자리에 점수를 준다

우리가 보고 싶은 것은 이것입니다.

> **다섯 명의 중간과 기말을 나란히 놓고 견주고 싶다.**

`hue=`가 "무리를 가르는 열"이니 중간과 기말을 가르려면 `hue`에 `"중간"`을 주면 될 것 같습니다. 빈칸 없음.

```python
fig2, ax2 = plt.subplots()
sns.barplot(data=df, x="이름", y="기말", hue="중간", ax=ax2)
ax2.set_title("Midterm vs final")
ax2.set_xlabel("Student")
ax2.set_ylabel("Score")

print("오류 없이 그려졌습니다. 범례가 하나 새로 생겼습니다.")
plt.show()
```

오류도 경고도 없습니다. 막대 다섯 개가 이름 순서대로 서 있습니다.
오른쪽에 **범례**가 새로 생겼는데, 제목 자리가 네모입니다(`hue`에 준 열 이름이 한글이라서요).

## Part E. 그런데 막대는 전부 맞습니다

`hue`를 주면 seaborn은 막대를 **무리별로 묶어** 그리기 때문에, `ax.patches`의 순서가 **범례 순서**를 따릅니다.
화면 왼쪽부터 보려면 `get_x()`로 한 번 정렬합니다. 빈칸 없음.

```python
bars = sorted([(p.get_x(), round(float(p.get_height()), 1)) for p in ax2.patches if p.get_height() > 0])
heights = [h for _, h in bars]
print("왼쪽부터 막대 높이:", heights)
print("기말 열          :", [float(v) for v in df["기말"]])
print("같은가:", heights == [float(v) for v in df["기말"]])
```

**틀린 값이 하나도 없습니다.** 그런데 이 그림은 **내가 물어본 것에 답하지 않습니다** —
중간 점수는 높이로 나오지 않았고 색이 되었을 뿐입니다.

`hue=`는 **무리를 가르는 이름표**를 받습니다. 값을 재는 자리는 `y=` 하나뿐입니다.

## Part F. 고치기 — 표의 모양을 바꾼다

문제는 `hue`가 아니라 **표의 모양**입니다. 지금 "중간이냐 기말이냐"는 **열 이름**에 있는데,

먼저 손으로 세어 보세요 — 다섯 명이 각각 두 시험을 봤으니 **긴 표의 줄은 몇 개**가 될까요?

- **빈칸 2:** `melt` 뒤의 **줄 수**

```python
guess_rows = ___                       # ✍️ 빈칸 2

long = df.melt(id_vars=["이름", "학과"], var_name="시험", value_name="점수")
print(long)
print("모양:", long.shape)
print("손으로 센 줄 수와 맞나:", guess_rows == long.shape[0])
```

## Part J. 🔹심화 — 점과 분포도 색으로

`hue=`는 막대에만 붙는 것이 아닙니다. V2의 산점도와 V3의 히스토그램에도 그대로 붙습니다.
V5에서 **칸을 둘로 나눠** 견주던 앞·뒤 보름을 오늘은 **한 칸에 색으로** 놓아 봅니다. 빈칸 없음.

```python
rentals = [14230, 15010, 13120, 13540, 6880, 12760, 13980, 16420, 15870, 13310,
           13650, 9240, 14020, 14890, 16060, 5120, 7340, 10210, 4050, 8760,
           12480, 5430, 5970, 11390, 9880, 11390, 4610, 7890, 13240, 13880]
temp = [24.1, 23.5, 22.8, 24.6, 21.2, 23.9, 25.3, 26.0, 26.4, 25.1,
        24.7, 22.9, 25.8, 26.6, 27.2, 23.1, 22.4, 24.0, 21.6, 23.3,
        26.1, 22.0, 21.8, 25.4, 24.2, 26.8, 22.6, 23.0, 27.5, 27.9]

bike = pd.DataFrame({"날짜": list(range(1, 31)), "대여건수": rentals, "평균기온": temp})
bike["반달"] = ["first half"] * 15 + ["second half"] * 15      # 무리를 가를 열을 먼저 만든다

fig9, (t1, t2) = plt.subplots(1, 2, figsize=(11, 4))
sns.scatterplot(data=bike, x="평균기온", y="대여건수", hue="반달", ax=t1)
sns.histplot(data=bike, x="대여건수", hue="반달", bins=6, ax=t2)
t1.set_title("Temperature vs rentals")
t2.set_title("Distribution of rentals")
t1.set_xlabel("Mean temp (C)")
t1.set_ylabel("Rentals")
t2.set_xlabel("Rentals")
t1.legend(title="Half")
t2.get_legend().set_title("Half")     # 🔴 히스토그램에는 ax.legend(title=) 를 쓰지 말 것 — 항목이 사라진다
fig9.tight_layout()

half = bike.groupby("반달")["대여건수"].mean()
print("반달별 평균:", list(half.index), [round(float(v), 1) for v in half.values])
plt.show()
```

V5의 **13532.0과 8776.0** 이 오늘은 색으로 갈립니다.
🔴 `hue`를 주려면 그 무리가 **열로 있어야** 하므로 `bike["반달"]` 열을 **먼저 만들었습니다** — `melt`와 같은 이야기입니다.

🔴 **두 그림의 범례 제목을 바꾸는 방법이 다릅니다.** 산점도는 `t1.legend(title="Half")`로 되지만,

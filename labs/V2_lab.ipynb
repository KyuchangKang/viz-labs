## Part F. 세어 보기 — x가 뒤로 간 횟수

"x가 앞 값보다 작아지는 구간의 수"를 세는 **함수**를 만듭니다(P7a). P3a의 누적기가 그 안에 들어갑니다.
30일 전체와 6/16~6/20 창(자리 15~19)을 같은 함수로 셉니다.

> **빈칸 4** — "이번 값이 **앞 값보다 작다**"를 뜻하는 비교 연산자를 쓰세요.

```python
def count_back(x):
    back = 0
    for i in range(1, len(x)):
        if x[i] ___ x[i - 1]:
            back = back + 1
    return back

back_all = count_back(temp)
back_week = count_back(temp[15:20])          # 6/16 ~ 6/20 다섯 값, 네 구간
print("6/16~6/20(네 구간)에서 뒤로 간 횟수:", back_week, "/ 내 손계산:", back_hand)
print("30일 전체에서 뒤로 간 횟수:", back_all, "/ 29 구간")
```

🔴 **데이터는 한 글자도 틀리지 않았습니다.** 선 그래프는 **행 순서**로 점을 잇는데, x를 기온으로 바꾸는 순간 **행 순서와 x 순서가 달라졌습니다.**
선은 30개 점 사이를 그만큼 **되돌아가며** 지나갑니다 — 그것이 실뭉치입니다.

## Part G. 고치기 ① — 기온 순으로 정렬한다

"그럼 기온 순으로 줄을 세우면 되겠네." P9의 `sorted`에 `key=`를 줍니다 — **자리 번호**(0~29)를 기온 순으로 정렬합니다. 빈칸 없음 — 설명서 §6.5와 같은 코드입니다.

```python
order = sorted(range(30), key=lambda i: temp[i])     # 기온이 낮은 날부터 자리 번호
temp_s = [temp[i] for i in order]
rent_s = [rentals[i] for i in order]

print("정렬 뒤 뒤로 간 횟수:", count_back(temp_s))
print("정렬 뒤 앞 셋 — 날:", [order[0] + 1, order[1] + 1, order[2] + 1])
print("기온:", temp_s[:3], "/ 대여:", rent_s[:3])

fig4, ax4 = plt.subplots()
ax4.plot(temp_s, rent_s)
ax4.set_xlabel("Temperature (C)")
ax4.set_ylabel("Rentals")
plt.show()
```

실뭉치는 풀렸습니다. 그런데 앞 셋을 보세요 — **6/5 → 6/19 → 6/23**. 이웃한 날도 아니고, 이어질 이유가 없는 세 날입니다.
그 선은 "기온이 21.2에서 21.6으로 오르면 대여가 6880에서 4050으로 준다"고 말합니다. **선이 뜻하는 "다음"이 없습니다.**

## Part J. 오류 읽기 — 그림 종류를 고를 때

일부러 네 가지 오류를 내 봅니다. **마지막 줄의 오류 이름과 메시지 앞부분**을 읽는 연습입니다. 빈칸 없음.

```python
fig7, ax7 = plt.subplots()

try:
    ax7.scatter(temp)
except TypeError as e:
    print("1)", type(e).__name__, "->", e)

try:
    ax7.plot(days, rentals[:29])
except ValueError as e:
    print("2)", type(e).__name__, "->", e)

try:
    ax7.plot(days, rentals, markers="o")
except AttributeError as e:
    print("3)", type(e).__name__, "->", e)

try:
    ax7.bar(weekdays, rentals[:7], marker="o")
except AttributeError as e:
    print("4)", type(e).__name__, "->", e)

plt.close(fig7)                              # 오류만 보고 그림은 닫는다
```

- 1)은 점에 **x만** 준 것, 2)는 x와 y의 **개수**가 다른 것(30과 29), 3)은 옵션 이름 오타(`marker`가 맞음), 4)는 **막대에는 `marker`가 없다**는 것.
- 오류가 **안 나는** 실수가 오늘의 세트피스입니다 — 순서 없는 x에 선을 긋는 것, 그리고 정렬해서 풀린 선을 믿는 것.

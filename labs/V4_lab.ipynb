```python
import matplotlib.pyplot as plt
import pandas as pd

rentals = [14230, 15010, 13120, 13540, 6880, 12760, 13980, 16420, 15870, 13310,
           13650, 9240, 14020, 14890, 16060, 5120, 7340, 10210, 4050, 8760,
           12480, 5430, 5970, 11390, 9880, 11390, 4610, 7890, 13240, 13880]
temp = [24.1, 23.5, 22.8, 24.6, 21.2, 23.9, 25.3, 26.0, 26.4, 25.1,
        24.7, 22.9, 25.8, 26.6, 27.2, 23.1, 22.4, 24.0, 21.6, 23.3,
        26.1, 22.0, 21.8, 25.4, 24.2, 26.8, 22.6, 23.0, 27.5, 27.9]
week = ["토", "일", "월", "화", "수", "목", "금"]          # 2024년 6월 1일은 토요일

df = pd.DataFrame({
    "날짜": list(range(1, 31)),
    "요일": [week[i % 7] for i in range(30)],
    "대여건수": rentals,
    "평균기온": temp,
})

print(df.shape)
print(df.head(3))
```

## Part B. 표에서 바로 그리기 — `ax=` 가 요점

## Part H. 조용히 틀리는 길

"이름표 순서로 다시 세우는 거면, 아예 **영어 이름표**를 주면 눈금까지 한꺼번에 해결되지 않을까?"

## Part K. 오류 읽기 — 판다스로 그릴 때

일부러 네 가지 오류를 내 봅니다. **오류 이름과 메시지 앞부분**을 읽는 연습입니다. 빈칸 없음.

```python
try:
    by_day.plot(kind="bars")
except ValueError as e:
    print("1)", type(e).__name__, "->", str(e)[:60])

try:
    df.plot(kind="bar", x="요일ㅁ")
except KeyError as e:
    print("2)", type(e).__name__, "->", e)

try:
    df["요일"].plot(kind="bar")
except TypeError as e:
    print("3)", type(e).__name__, "->", e)

try:
    df.plot(kind="scatter", x="평균기온")
except ValueError as e:
    print("4)", type(e).__name__, "->", e)

plt.close("all")                             # 오류만 보고 그림은 닫는다
```

- 1)은 `kind`의 철자(`"bar"` 단수), 2)는 표에 없는 **열 이름**, 3)은 **글자 열**을 그리려 한 것, 4)는 `scatter`에 `y=`를 안 준 것.
- 오늘의 세트피스는 **오류가 안 나는** 쪽입니다 — 눈금 글자만 갈아 끼우는 것.

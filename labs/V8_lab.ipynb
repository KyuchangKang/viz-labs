```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

rentals = [14230, 15010, 13120, 13540, 6880, 12760, 13980, 16420, 15870, 13310,
           13650, 9240, 14020, 14890, 16060, 5120, 7340, 10210, 4050, 8760,
           12480, 5430, 5970, 11390, 9880, 11390, 4610, 7890, 13240, 13880]
temp = [24.1, 23.5, 22.8, 24.6, 21.2, 23.9, 25.3, 26.0, 26.4, 25.1,
        24.7, 22.9, 25.8, 26.6, 27.2, 23.1, 22.4, 24.0, 21.6, 23.3,
        26.1, 22.0, 21.8, 25.4, 24.2, 26.8, 22.6, 23.0, 27.5, 27.9]
week = ["토", "일", "월", "화", "수", "목", "금"]          # 2024년 6월 1일은 토요일

df = pd.DataFrame({
    "날짜": list(range(1, 31)),
    "요일": [week[i % 7] for i in range(30)],
    "대여건수": rentals,
    "평균기온": temp,
})

order = ["월", "화", "수", "목", "금", "토", "일"]
eng = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

print("모양:", df.shape)
print("전체 평균:", round(float(df["대여건수"].mean()), 1))
print("seaborn 판:", sns.__version__)
```

> ⚠️ **이 셀부터 경고가 하나 함께 나옵니다** — `UserWarning: set_ticklabels() should only be used with a fixed number of ticks`.
> seaborn의 x축은 판다스 `.plot`과 눈금을 잡는 방식이 달라서 붙는 것이고, **`order=`로 자리를 이미 고정했으므로 이름표는 제대로 붙습니다.**
> 그림에는 문제가 없으니 놀라지 마세요.

```python
figB, axB = plt.subplots(figsize=(7, 4))
sns.barplot(data=df, x="요일", y="대여건수", order=order, errorbar=None, ax=axB)
axB.set_xticklabels(eng)                 # order 로 순서를 정한 뒤이므로 안전하다
axB.set_title("Average rentals by weekday")
axB.set_xlabel("Weekday")
axB.set_ylabel("Rentals")
figB.tight_layout()
plt.show()

print("막대 높이:", [round(float(p.get_height()), 1) for p in axB.patches])
print("x 눈금:", [t.get_text() for t in axB.get_xticklabels()])
```

## Part F. 그런데 두 번 그리면 값이 달라집니다

```python
runs = []
for _ in range(5):
    figF, axF = plt.subplots()
    sns.barplot(data=df, x="요일", y="대여건수", order=order, ax=axF)
    runs.append(tuple([round(float(ln.get_ydata()[0]), 1) for ln in axF.lines]))   # 집합에 넣으려면 리스트가 아니라 tuple 이어야 한다
    plt.close(figF)

print("다섯 번 그렸습니다.")
print("다섯 번이 모두 같은가:", len(set(runs)) == 1)
```

🔴 기본값은 **`errorbar=("ci", 95)`** — 95% 신뢰구간이고, 그 구간을 **부트스트랩**으로 구합니다.

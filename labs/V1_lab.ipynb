## Part J. 오류 읽기 — 그림을 그릴 때

일부러 네 가지 오류를 내 봅니다. **마지막 줄의 오류 이름과 메시지 앞부분**을 읽는 연습입니다. 빈칸 없음.

```python
fig6, ax6 = plt.subplots()

try:
    ax6.title("Final Scores")
except TypeError as e:
    print("1)", type(e).__name__, "->", e)

try:
    ax6.ylabel("Score")
except AttributeError as e:
    print("2)", type(e).__name__, "->", e)

try:
    plt.set_title("Final Scores")
except AttributeError as e:
    print("3)", type(e).__name__, "->", e)

try:
    ax6.bar(labels)
except TypeError as e:
    print("4)", type(e).__name__, "->", e)

plt.close(fig6)                              # 오류만 보고 그림은 닫는다
```

- 1)·2)는 `ax.` 문법에 **`set_` 을 빼먹은** 것, 3)은 반대로 `plt.` 문법에 **`set_` 을 붙인** 것입니다.
- 4)는 `bar` 에 **높이를 안 준** 것입니다.
- 오류가 **안 나는** 실수도 있습니다 — 한글 축 이름(`□□□`), `label=` 없이 범례를 부르는 것(빈 범례), 그리고 **오늘 세트피스**.

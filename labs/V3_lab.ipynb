## Part E. 봉우리 세는 함수

"봉우리가 몇 개인가"를 눈짐작으로 말하면 사람마다 다릅니다. 규칙을 정합니다.

> 어떤 칸이 **왼쪽 칸보다 높고**, **오른쪽 칸보다 낮지 않으면** 봉우리로 센다.
> 양 끝 바깥에는 높이 **0**인 칸이 있다고 친다. 같은 높이가 이어지면 **왼쪽 첫 칸**만 센다.

V2의 `count_back`과 같은 꼴입니다 — P7a의 함수 안에 P3a의 누적기가 들어갑니다.

> **빈칸 4** — "**오른쪽 칸보다 낮지 않다**"를 뜻하는 비교 연산자를 쓰세요.

```python
def count_peaks(counts):
    padded = [0] + [int(c) for c in counts] + [0]      # 양 끝에 0인 칸을 붙인다 (PD1: 리스트 이어붙이기)
    peaks = 0
    for i in range(1, len(padded) - 1):
        if padded[i] > padded[i - 1] and padded[i] ___ padded[i + 1]:
            peaks = peaks + 1
    return peaks

peaks5 = count_peaks(counts5)
print("bins=5 의 봉우리:", peaks5)
print("규칙 시험 — [1, 2, 2, 1] 의 봉우리:", count_peaks([1, 2, 2, 1]))   # 같은 높이가 이어지면 왼쪽 첫 칸만
```

## Part J. 오류 읽기 — 칸을 잘못 정할 때

일부러 네 가지 오류를 내 봅니다. **오류 이름과 메시지 앞부분**을 읽는 연습입니다. 빈칸 없음.

```python
fig10, ax10 = plt.subplots()

try:
    ax10.hist(rentals, bin=5)
except AttributeError as e:
    print("1)", type(e).__name__, "->", e)

try:
    ax10.hist(rentals, bins=0)
except ValueError as e:
    print("2)", type(e).__name__, "->", e)

try:
    ax10.hist(rentals, bins=[4000, 8000, 6000])
except ValueError as e:
    print("3)", type(e).__name__, "->", e)

try:
    ax10.hist(rentals, bins=5.5)
except TypeError as e:
    print("4)", type(e).__name__, "->", e)

plt.close(fig10)                             # 오류만 보고 그림은 닫는다
```

- 1)은 옵션 이름 오타(복수형이어야 합니다), 2)는 칸 수가 0, 3)은 경계 목록이 **커지는 순서**가 아닌 것, 4)는 칸 수에 **소수**를 준 것.
- 오늘의 세트피스는 **오류가 안 나는** 쪽입니다 — 칸 수를 바꿔 놓고 그림만 읽는 것.
